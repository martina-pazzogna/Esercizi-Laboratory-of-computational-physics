6\. Use broadcasting to create a grid of distances

Route 66 crosses the following cities in the US: Chicago, Springfield, Saint-Louis, Tulsa, Oklahoma City, Amarillo, Santa Fe, Albuquerque, Flagstaff, Los Angeles
The corresponding positions in miles are: 0, 198, 303, 736, 871, 1175, 1475, 1544, 1913, 2448

  * Construct a 2D grid of distances among each city along Route 66
  * Convert that in km (those savages...)

```python
import numpy as np
positions_miles = np.array([0, 198, 303, 736, 871, 1175, 1475, 1544, 1913, 2448])
positions_reshaped = positions_miles.copy()
positions_reshaped = positions_reshaped.reshape((len(positions_miles),1))
distances_miles = abs(positions_miles - positions_reshaped)
print('The grid of distances in miles among each city along Route 66 is \n', distances_miles)
```

```python
conv = 1.609344
distances_km = distances_miles / conv
print("The grid of distances in km among each city along Route 66 is \n", distances_km)
```
