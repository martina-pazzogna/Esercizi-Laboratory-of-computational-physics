2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
#converts the mantissa
def mantissa_conversion(mant):
    x = 1
    for i in range(0, len(mant)):
        x += mant[i] * 2 **(-(i+1))
    
    return x
    
    
```

```python
#converts an integer from binary to decimal
def bindec_i(bin):
    
    dec = 0
    
    for i in range(0, len(bin)):
        dec += bin[i] * 2**(len(bin) - 1 - i)
        
    return dec
```

```python
#converts a real number from binary to decimal
def real_conversion(binary):
    
    s = binary[0]
    mantissa = mantissa_conversion(binary[9:32])
    e = bindec_i(binary[1:9])
    
    real = (-1)**s * mantissa * 2**(e-127)
    
    return real
```
