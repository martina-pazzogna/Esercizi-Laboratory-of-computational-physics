5\. Create a matrix that shows the 10 by 10 multiplication table.

 * Find the trace of the matrix
 * Extract the anto-diagonal (this should be ```array([10, 18, 24, 28, 30, 30, 28, 24, 18, 10])```)
 * Extract the diagnoal offset by 1 upwards (this should be ```array([ 2,  6, 12, 20, 30, 42, 56, 72, 90])```)

````python
import numpy as np
m = np.array([[i*j for j in range(1,11)] for i in range(1,11)] )
trace = np.trace(m)
anti_diagonal = np.diag(m[:, ::-1])
upper_diagonal = np.diag(m, +1)
print("The 10 by 10 multiplication table is \n", m)
print("The trace of this matrix is ", trace)
print("The anti-diagonal of this matrix is \n", anti_diagonal)
print("The upper diagonale of this matrix is \n", upper_diagonal)
````
