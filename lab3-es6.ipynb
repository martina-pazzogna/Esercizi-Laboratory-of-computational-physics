6\. Write a program that implements the function $f(x)=x(x−1)$

(a) Calculate the derivative of the function at the point $x = 1$ using the derivative definition:

$$
\frac{{\rm d}f}{{\rm d}x} = \lim_{\delta\to0} \frac{f(x+\delta)-f(x)}{\delta}
$$

with $\delta = 10^{−2}$. Calculate the true value of the same derivative analytically and compare with the answer your program gives. The two will not agree perfectly. Why not?

(b) Repeat the calculation for $\delta = 10^{−4}, 10^{−6}, 10^{−8}, 10^{−10}, 10^{−12}$ and $10^{−14}$. How does the accuracy scales with $\delta$?

```python
def f(x):
    #returns x^2-x
    return x**2 - x
```

```python
def derivative_delta(x, delta):
    #returns the derivative of f(x) using the derivative definition

    y = (f(x + delta) - f(x))/delta

    return y
```

```python
def derivative(x):
    #returns the analytical derivative of f(x)

    return 2*x - 1
```

```python
y1 = derivative_delta(1, 10**(-2))
y2 = derivative(1)
print("The derivative of f(x) at point x=1 calculated using the derivative definition with $\delta=10^{-2}$ is", y1)
print("The derivative of f(x) at point x=1 calculated using the analytical derivative is", y2)
```

When we calculate the derivative of f(x) at x=1 using the derivative definition, the numerator is obtained as the difference bewteen two close numbers and this can frequently cause issues. We expect that, the smaller is $\delta$, the biggest is the inaccuracy in the final result.

```python
exp = [4,6,8,10,12,14]
derivatives = [derivative_delta(1, 10**(-e)) for e in exp]
print(derivatives)
```

```python
inaccuracy = [ abs(1.0 - x) for x in derivatives]
inaccuracy
```
