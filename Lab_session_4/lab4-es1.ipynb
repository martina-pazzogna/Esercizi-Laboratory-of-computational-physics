1\.a Make a new directory called `students` in your home. Download a csv file with the list of students of this lab from [here](https://www.dropbox.com/s/867rtx3az6e9gm8/LCP_22-23_students.csv) (use the `wget` command) and copy that to `students`. First check whether the file is already there

1\.b Make two new files, one containing the students belonging to PoD, the other to Physics.

1\.c For each letter of the alphabet, count the number of students whose surname starts with that letter.

1\.d Find out which is the letter with most counts.

1\.e Assume an obvious numbering of the students in the file (first line is 1, second line is 2, etc.), group students "modulo 18", i.e. 1,19,37,.. 2,20,38,.. etc. and put each group in a separate file  

```python
#!/bin/bash

#creating the directory, dowloading the file and copying it to the new directory
mkdir students

wget https://www.dropbox.com/scl/fi/bxv17nrbrl83vw6qrkiu9/LCP_22-23_students.csv?rlkey=47fakvatrtif3q3qw4q97p5b7&e=1

if [ ! -f "./students/LCP_22-23_students.csv\?rlkey\=47fakvatrtif3q3qw4q97p5b7" ]
then 
    cp LCP_22-23_students.csv\?rlkey\=47fakvatrtif3q3qw4q97p5b7 ./students/LCP_22-23_students.csv\?rlkey\=47fakvatrtif3q3qw4q97p5b7
fi

cd students

#dividing the students from Physics of Data from the ones from Physics and putting they data into different files
grep "PoD" LCP_22-23_students.csv\?rlkey\=47fakvatrtif3q3qw4q97p5b7 

touch PoD_students.csv

nano PoD_students.csv

grep "Physics" "LCP_22-23_students.csv\?rlkey\=47fakvatrtif3q3qw4q97p5b7"

touch Physics_students.csv

nano Physics_students.csv

#counting how many students' surnames start with each letter of the alphabet and finding out the one with most counts
max=0
letter=A
for l in {a..z}
    do
    counter=0
    while IFS= read -r r
        do
        first=$(echo "$r" | cut -d '.' -f 1 | cut -c 1 | tr '[:upper:]' '[:lower:]')
        if [[ "$first" == "$l" ]]
        then
            ((counter++))
        fi
    done < <(tail -n +2 LCP_22-23_students.csv\?rlkey\=47fakvatrtif3q3qw4q97p5b7)
    echo "$l: $counter"
    if [[ "$counter" -gt "$max" ]]
        then
            letter=$l
            max=$counter
    fi
done
echo "The letter with most counts is $letter"     

#grouping the students "modulo 18" and putting each group in a separate file
for g in {1..18}
do
    touch group_${g}.csv
    n=$g
    rows=$(tail -n +2 LCP_22-23_students.csv\?rlkey\=47fakvatrtif3q3qw4q97p5b7 | wc -l)

    while [[ $n -le $rows ]]
    do
        sed -n "$((n+1))p" LCP_22-23_students.csv\?rlkey\=47fakvatrtif3q3qw4q97p5b7 >> group_${g}.csv
        ((n+=18))
    done
done
```
