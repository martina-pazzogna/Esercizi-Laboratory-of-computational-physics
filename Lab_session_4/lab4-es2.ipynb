2.a Make a copy of the file `data.csv` removing the metadata and the commas between numbers; call it `data.txt`

2\.b How many even numbers are there?

2.c Distinguish the entries on the basis of sqrt(X^2 + Y^2 + Z^2) is greater or smaller than 100*sqrt(3)/2. Count the entries of each of the two groups

2\.d Make `n` copies of data.txt (with `n` an input parameter of the script), where the i-th copy has all the numbers divided by i (with `1<=i<=n`).

```python
#!/bin/bash

#creating the file data.csv in a directory called data
mkdir data
cd data
touch data.csv
nano data.csv #writing the file

#removing the metadata and the commas in the copy called data.txt
cp data.csv data.txt
sed -i '/^#/d' data.txt
sed -i 's/,//g' data.txt

#counting the even numbers
even_counter=0
for x in $(cat data.txt) #we use data.txt because cat works well on files whose elements are separated by blank spaces
do
    if [[ $((x % 2)) -eq 0 ]]
    then
    ((even_counter++))
    fi
done
echo "The number of even numbers in data.txt is $even_counter$"

#diving the entries using a threshold
threshold=$(echo "scale=6; 100 * sqrt(3) / 2" | bc -l)
bigger=0
smaller=0
while IFS= read -r r
do
    x=$(echo "$r" | cut -d ',' -f 1)
    y=$(echo "$r" | cut -d ',' -f 2)
    z=$(echo "$r" | cut -d ',' -f 3)

    if [[ $(echo "($x)^2 + ($y)^2 + ($z)^2 >= ($threshold)^2" | bc -l) -eq 1 ]]
    then
        ((bigger++))
    else
        ((smaller++))
    fi
done < <(tail -n +5 data.csv)
echo "The number of entries with sqrt(X^2 + Y^2 + Z^2) greater than $threshold is $bigger."

#creating n copies of data.txt such that the i-th copy has all entries divided by i
n=$1
for i in $(seq 1 $n)
do
    > temp.txt
    while IFS= read -r r
    do
        for x in $r
        do
            risultato=$(echo "scale=6; $x/$i" | bc -l)
            printf "%s " "$risultato" >> temp.txt

        done
        
        printf "\n" >> temp.txt

    done <(tail -n +1 data.txt)

cp temp.txt data_${i}.txt

done


    
    
        
        
```
