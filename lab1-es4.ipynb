4. Counting letters

Write a program to calculate the number of times each character occurs in a given string s. Ignore differneces in capitalization

```python
def counter(s):
    s_low = s.lower()
    char = {}
    
    for x in s_low:
        counter = 0
        for y in s_low:
            if (x == y):
                counter += 1
        char[x] = counter

    return char
    
```

```python
word = "parola"
characters = counter(word)
print(characters)
```
