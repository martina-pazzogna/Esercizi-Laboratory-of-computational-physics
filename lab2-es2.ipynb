2. Convert the following function into a pure function with no global variables or side effects

```python
x = 5
def f(alist):
    for i in range(x):
         alist.append(i)
    return alist

alist = [1,2,3]
ans = f(alist)
print (ans)
print (alist) # alist has been changed!
```

```python
import copy
x = 5
def f_pure(alist):
    newlist = alist.copy()
    for i in range(x):
        newlist.append(i)

    return newlist

alist1 = [1,2,3]
ans1 = f_pure(alist1)
print (ans1)
print (alist1)
```
