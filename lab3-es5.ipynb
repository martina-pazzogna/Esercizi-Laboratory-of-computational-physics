5\. Write a function that takes in input three parameters $a$, $b$ and $c$ and prints out the two solutions to the quadratic equation $ax^2+bx+c=0$ using the standard formula:
$
x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}
$

(a) use the program to compute the solution for $a=0.001$, $b=1000$ and $c=0.001$

(b) re-express the standard solution formula by multiplying top and bottom by $-b\mp\sqrt{b^2-4ac}$ and again find the solution for $a=0.001$, $b=1000$ and $c=0.001$. How does it compare with what previously obtained? Why?

(c) write a function that compute the roots of a quadratic equation accurately in all cases

```python
import math
def solutions(a,b,c):

    if (b**2 >= 4 * a * c):
        s_r = math.sqrt(b**2 - 4 * a * c)
        den = 1 / (2 * a)

        x1 = (-b + s_r) * den
        x2 = (-b - s_r) * den

        epsilon = 0.000000000000000000000000000000000001
        if (abs(x1-x2) <= epsilon):
            print("The given equation has two coincident solutions which are x1=x2=", x1)
        else:
            print("The two solutions to the given equation are x1=", x1, "and x2=", x2)
    
    else: 
        print("There are no real solutions to the given equation")
```

```python
solutions(1,2,1)
```

```python
solutions(1,2,5)
```

```python
solutions(0.001, 1000, 0.001)
```

```python
#new implementation of the function after re-expressing the standard formula
def new_solutions(a,b,c):

    if (b**2 >= 4 * a * c):
        den1 = -b - math.sqrt(b**2 - 4 * a * c)
        den2 = -b + math.sqrt(b**2 - 4 * a * c)

        x1 = 2 * c/den1
        x2 = 2 * c/den2

        epsilon = 0.000000000000000000000000000000000001
        if (abs(x1-x2) <= epsilon):
            print("The given equation has two coincident solutions which are x1=x2=", x1)
        else:
            print("The two solutions to the given equation are x1=", x1, "and x2=", x2)
    
    else: 
        print("There are no real solutions to the given equation")
```

```python
new_solutions(0.001, 1000, 0.001)
```

In the first case, when calculating $-b- \sqrt{b^2 - 4ac}$, the system is subtracting two numbers which are almost the same.
In the second case, this problem refers to the denominator of the bigger solution.
We can solve this problem by calculating the two solutions adopting the fisrt expression of the standard formula in one case and the re-expressed expression in the other

```python
def solutions_correct(a,b,c):
    
    if (b**2 >= 4 * a * c):
        x1 = 2 * c / (- b - math.sqrt(b**2 - 4 * a * c))
        x2 = (- b - math.sqrt(b**2 - 4 * a * c)) / (2 * a)

        epsilon = 0.000000000000000000000000000000000001
        if (abs(x1-x2) <= epsilon):
            print("The given equation has two coincident solutions which are x1=x2=", x1)
        else:
            print("The two solutions to the given equation are x1=", x1, "and x2=", x2)
    
    else: 
        print("There are no real solutions to the given equation")
```
