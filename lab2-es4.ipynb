```python
def factorial(x):
    i = 1
    for j in range(1,x+1):
        i = i * j
    return i
```

```python
x = factorial(4)
print(x)
```

```python
from functools import reduce
def mult(x,y):
    return x * y
def fac_no_rec(x):
    return reduce(mult, range(1, x+1))
```

```python
y = fac_no_rec(6)
print(y)
```
