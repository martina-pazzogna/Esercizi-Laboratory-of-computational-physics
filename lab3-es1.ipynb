1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
#converts an integer from decimal to binary
def decbin_i(n):
    
    binary = []
    num = n
    
    while num>0:
        if(num%2 == 0):
            binary.append(0)
        else:
            binary.append(1)
            
        num  = int(num/2)
        
    binary.reverse()
    
    return binary
```

```python
a = decbin(30)
a
```

```python
#converts an integer from binary to decimal
def bindec_i(bin):
    
    dec = 0
    
    for i in range(0, len(bin)):
        dec += bin[i] * 2**(len(bin) - 1 - i)
        
    return dec
```

```python
x = bindec_i([1,1,1,1,0])
x
```

```python
#converts an integer from decimal to hexadecimal
def dechex_i(n):
    
    hexa = []
    num = n
    
    while num > 0:
        carry = num % 16
        if (carry == 15):
            hexa.append('F')
        elif (carry == 14):
            hexa.append('E')
        elif(carry == 13):
            hexa.append('D')
        elif(carry == 12):
            hexa.append('C')
        elif(carry == 11):
            hexa.append('B')
        elif(carry == 10):
            hexa.append('A')
        else:
            hexa.append(carry)
        
        num = int(num/16)
        
    hexa.reverse()
    return hexa
```

```python
b = dechex_i(30)
b
```

```python
#converts an integer from hexadecimal to decimal
def hexdec_i(hexa):
    
    dec = 0
    
    for i in range(0,len(hexa)):
        dig = hexa[i]
        if(dig == 'F'):
            dec += 15 * 16**(len(hexa) - 1 - i)
        elif(dig == 'E'):
            dec += 14 * 16**(len(hexa) - 1 - i)
        elif(dig == 'D'):
            dec += 13 * 16**(len(hexa) - 1 - i)
        elif(dig == 'C'):
            dec += 12 * 16**(len(hexa) - 1 - i)
        elif(dig == 'B'):
            dec += 11 * 16**(len(hexa) - 1 - i)
        elif(dig == 'A'):
            dec += 10 * 16**(len(hexa) - 1 - i)
        else:
            dec += dig * 16**(len(hexa) - 1 - i)
    
    return dec
```

```python
y = hexdec_i([1, 'E'])
y
```
