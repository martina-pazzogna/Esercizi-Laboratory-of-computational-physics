````python
import numpy as np
import matplotlib.pyplot as plt
````

````python
data = np.loadtxt("populations.txt", skiprows=1)
year = data[:,0]
hares = data[:,1]
lynxes = data[:,2]
carrots = data[:,3]
````

````python
plt.plot(year, hares, color = 'Green', label="Hares' population")
plt.plot(year, lynxes, color = 'Blue', label = "Lynxes' population")
plt.plot(year, carrots, color = 'Orange', label = "Carrots availability")
plt.xlabel('Year')
plt.ylabel('Population')
plt.legend(loc = 'best')
plt.show()
````

````python
means = np.array([np.mean(data[:,i]) for i in range(1, data.shape[1])])
stds = np.array([np.std(data[:,i]) for i in range(1, data.shape[1])])
r_h_c = np.corrcoef(hares, carrots)[0,1]
r_l_c = np.corrcoef(lynxes, carrots)[0,1]
r_h_l = np.corrcoef(lynxes, hares)[0,1] 
print('The correlation coefficient between hares and carrots is: ', r_h_c)
print('The correlation coefficient between lynxes and carrots is: ', r_l_c)
print('The correlation coefficient between hares and lynxes is: ', r_h_l)
````

````python
highest_pop = []
for i in range(len(year)):
    idx = np.argmax(data[i,1:]) + 1 
    if (idx == 1):
        highest_pop.append('Hares')
    if (idx == 2):
        highest_pop.append('Lynxes')
    if( idx == 3):
        highest_pop.append('Carrots')
    
highest_pop = np.asarray(highest_pop)
print('The species with the highest population each year are \n', highest_pop)
````
