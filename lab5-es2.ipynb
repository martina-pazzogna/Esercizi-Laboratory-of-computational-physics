2\. Find the outer product of the following two vecotrs

```python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
```

Do this in the following ways:

   * Using the function outer in numpy
   * Using a nested for loop or list comprehension
   * Using numpy broadcasting operatoins

````python
import numpy as np
````

````python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
````

````python
p1 = np.outer(u,v)
print('The outer product obtained using np.outer() is \n', p1)
````

````python
p2 = np.array([[u[i] * v[j] for j in range(len(v))] for i in range(len(u))])
print('The outer product calculated using list comprehension is \n', p2)
````

````python
u1 = u.reshape((4,1))
p3 = u1 * v
print('The outer product calculated using numpy broadcasting operations is \n', p3)
````
