1. The MickeyMouse problem

a) Write a program that prints the numbers from 1 to 100. But for multiples of 3 print `Mickey` instead of the corresponding number and for the multiples of 5 print `Mouse`. For numbers which are multiples of both three and five print `MickeyMouse`

b) Put the result in a tuple and substitute `Mickey` with `Donald` and `Mouse` with `D

```python
result = ()

for i in range(1,101):
    element = 0
    
    if ((i%3 == 0) & (i%5 != 0)):
        element = "Mickey"
        print(element)
      
    elif((i%3 !=0) & (i%5 == 0)):
        element = "Mouse"
        print(element)

    elif((i%3 == 0) & (i%5 == 0)):
        element = "MickeyMouse"
        print(element)
    else: 
        element = i
        print(element)


    result += (element,)
```

```python
result_list = list(result)
for i in range(0, len(result_list)):
    if (result_list[i] == "Mickey"):
        result_list[i] = "Donald"

    elif (result_list[i] == "Mouse"):
        result_list[i] = "D"
result_mod = tuple(result_list)
print(result_mod)
```
