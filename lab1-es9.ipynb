9\. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one

```python
import math
def normalization(t):
    partial = 0
    for x in t:
        partial += x**2
    norm = math.sqrt(partial)
    normalized = tuple( x/norm for x in t)
    return normalized
```

```python
numbers = (1,4,0)
norm = normalization(numbers)
print(norm)
```
