2. The swap function

Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable

```python
def swap(x,y):
    temp = x 
    x = y
    y = temp

    return x,y
    
```

```python
a,b = swap(1,3)
print(a, ",", b)
```

```python
def swap_no_temp(x,y):
    list = [x,y]
    a  = list[1]
    b = list[0]

    return a,b
```

```python
c,d = swap_no_temp(4,8)
print(c, ",", d)
```
