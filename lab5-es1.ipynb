1\. Find the row, column and overall means for the following matrix:

```python
m = np.arange(12).reshape((3,4))
```

````python
import numpy as np
m = np.arange(12).reshape((3,4))
mean_rows = m.mean(axis=1)
mean_columns = m.mean(axis=0)
mean_tot = m.mean()
print('The means along the rows are ', mean_rows)
print('The means along the columns are', mean_columns)
print('The overall mean is ', mean_tot)
````
