3\. Create a 10 by 6 matrix of random uniform numbers. Set all rows with any entry less than 0.1 to be zero

Hint: Use the following numpy functions - np.random.random, np.any as well as Boolean indexing and the axis argument

```python
import numpy as np
```

```python
m = np.random.random((10,6))
mask = (m < 0.1) 
m_masked = np.any(mask, axis=1, keepdims= True)
idx = np.array([i for i in range(m_masked.shape[0]) if m_masked[i][0] == True])
m_new = m.copy()
m_new[idx, :]=0
print('The original matrix is \n', m)
print('The matrix with rows set to 0 if any of its entry is less than 0.1 is \n', m_new)
```
